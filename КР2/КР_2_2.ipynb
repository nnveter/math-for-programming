# Контрольная работа № 2.2. Обыкновенные дифференциальные уравнения
Ответы на вопросы из `Математика для программирования КР 2.2.pdf`.

**1. Что такое ОДУ?** Уравнение, связывающее функцию одной переменной и её производные: $F(x,y,y',\dots,y^{(n)})=0$. Пример: $y'=-2y$.

**2. Что такое порядок?** Наибольший порядок производной, входящей в уравнение. Пример второго порядка: $y''+4y=0$.

**3. Чем общее решение отличается от частного?** Общее содержит произвольные постоянные (столько, каков порядок уравнения) и описывает всё семейство решений. Частное получается из общего при конкретных значениях постоянных, например из начальных условий. Для $y'=-2y$: общее $y=Ce^{-2x}$, частное при $y(0)=1$: $y=e^{-2x}$.

**4. Задача Коши.** Найти решение уравнения $y'=f(x,y)$, удовлетворяющее начальному условию $y(x_0)=y_0$. Для порядка $n$ задаются $y,y',\dots,y^{(n-1)}$ в одной точке $x_0$.

**5. Поле направлений.** Каждой точке $(x,y)$ сопоставляется отрезок с наклоном $f(x,y)$. Интегральная кривая в каждой своей точке касается отрезка поля: решение идёт вдоль поля.

**6. Теорема существования и единственности.** Если $f$ и $\partial f/\partial y$ непрерывны в окрестности точки $(x_0,y_0)$, то в некоторой окрестности $x_0$ решение задачи Коши существует и единственно. Через точку проходит ровно одна интегральная кривая, поэтому численно искать решение имеет смысл.

**7. Разделяющиеся переменные.** Вид $y'=f(x)g(y)$. Решение: $\dfrac{dy}{g(y)}=f(x)\,dx$, затем интегрирование обеих частей. Отдельно проверяются решения $g(y)=0$, потерянные при делении.

**8. Линейное первого порядка.** Вид $y'+p(x)y=q(x)$. Методы: интегрирующий множитель или вариация постоянной.

**9. Интегрирующий множитель.** Функция $\mu=e^{\int p\,dx}$. После умножения на неё левая часть становится производной произведения: $(\mu y)'=\mu q$, и уравнение решается одним интегрированием.

**10. Однородное уравнение.** Вид $y'=f(y/x)$. Замена $y=ux$, $y'=u+xu'$ приводит к уравнению с разделяющимися переменными.

**11.** $y'=xy$: $\dfrac{dy}{y}=x\,dx$, значит $y=Ce^{x^2/2}$.

**12.** $y'+y=e^x$: $\mu=e^x$, $(e^xy)'=e^{2x}$, значит $y=\dfrac12e^x+Ce^{-x}$.

**13. Нормальная форма системы.** Система, разрешённая относительно производных: $\mathbf y'=\mathbf f(x,\mathbf y)$, $\mathbf y\in\mathbb R^n$.

**14. Сведение $y''=f(x,y,y')$.** Пусть $y_1=y$, $y_2=y'$. Тогда $y_1'=y_2$, $y_2'=f(x,y_1,y_2)$.

**15. Маятник.** Пусть $\omega=\theta'$. Тогда $\theta'=\omega$, $\omega'=-\dfrac gL\sin\theta$.

**16. Зачем сводить к системам?** Численные методы и библиотеки (например, `solve_ivp`) работают с системами первого порядка. Один и тот же код решает уравнения любого порядка: меняется только правая часть.

**17. Метод Эйлера.** Шаг вдоль касательной: $y_{i+1}=y_i+hf(x_i,y_i)$. Это ряд Тейлора, обрезанный на первой производной.

**18. Метод Хойна.** Прогноз методом Эйлера, затем усреднение наклонов в начале и в конце шага: $y_{i+1}=y_i+\dfrac h2(k_1+k_2)$, где $k_1=f(x_i,y_i)$, $k_2=f(x_i+h,y_i+hk_1)$. Это метод трапеций для интеграла.

**19. RK4.** Вычисляются четыре наклона: в начале, дважды в середине и в конце шага. Они усредняются с весами $1,2,2,1$: $y_{i+1}=y_i+\dfrac h6(k_1+2k_2+2k_3+k_4)$. Это аналог формулы Симпсона.

**20. Локальная и глобальная погрешность.** Локальная — ошибка за один шаг при точном начальном значении. Глобальная — накопленная ошибка на всём отрезке. Если локальная равна $O(h^{p+1})$, то глобальная равна $O(h^p)$.

**21. Порядки точности (глобальные).** Эйлер — $1$, Хойн — $2$, RK4 — $4$.

**22. Как проверить порядок экспериментально?** Вычислить ошибку при нескольких шагах $h$ на задаче с известным решением и построить график в лог-лог масштабе: наклон прямой равен порядку. Либо сравнить ошибки при $h$ и $h/2$: отношение равно примерно $2^p$.

**23. Устойчивость.** Способность метода не усиливать возмущения и ошибки округления: на затухающей задаче численное решение тоже должно затухать при заданном шаге.

**24. Условие устойчивости Эйлера на $y'=\lambda y$.** $|1+h\lambda|<1$. Для вещественного $\lambda<0$ это $h<2/|\lambda|$.

**25. Жёсткая система.** Система, в которой есть сильно различающиеся временные масштабы: $|\lambda_{\max}|/|\lambda_{\min}|\gg1$ при отрицательных действительных частях собственных значений матрицы Якоби. Пример: $y_1'=-1000y_1$, $y_2'=-y_2$.

**26. Почему явные методы плохо работают на жёстких задачах?** Их шаг ограничен устойчивостью по самой быстрой компоненте ($h\lesssim1/|\lambda_{\max}|$), даже когда она уже затухла. Шагов получается очень много. Нужно использовать неявные A-устойчивые методы: неявный Эйлер, BDF, Radau.

**27. Краевая задача и задача Коши.** В задаче Коши все условия заданы в одной точке, в краевой — на концах отрезка, например $y(a)=\alpha$, $y(b)=\beta$. Начальной производной нет, поэтому нельзя сразу применить RK4, а решение может быть не единственным или отсутствовать.

**28. Метод стрельбы.** Неизвестное $y'(a)=s$ делают параметром, решают задачу Коши и подбирают $s$ так, чтобы $y(b;s)=\beta$. Подбор — это поиск корня функции, например методами бисекции или Брента.

**29. Метод конечных разностей.** На сетке производные заменяют разностями: $y''(x_i)\approx\dfrac{y_{i-1}-2y_i+y_{i+1}}{h^2}$. Получается система алгебраических уравнений. Для линейной задачи она трёхдиагональная и решается прогонкой за $O(N)$. Порядок точности равен $2$.

**30. Методы `solve_ivp`.**
| Метод | Тип | Когда использовать |
|---|---|---|
| `RK45` | явный, порядок 5(4), по умолчанию | нежёсткие задачи общего вида |
| `RK23` | явный, порядок 3(2) | низкая точность, дешёвый шаг |
| `DOP853` | явный, порядок 8 | нежёсткие задачи с высокой точностью |
| `BDF` | неявный многошаговый, порядок 1–5 | типичные жёсткие задачи |
| `Radau` | неявный РК, порядок 5 | очень жёсткие задачи, высокая надёжность |
| `LSODA` | автопереключение Адамс/BDF | когда характер задачи заранее неизвестен |
